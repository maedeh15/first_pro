# run75 — priority-DECFUSION (STA-Net input-dependent decision fusion)

Replaces the three fixed sigmoid weights (Eq. 4) by per-sample softmax weights computed from each branch classifier's **hidden features** (STA-Net Eq. 5-6). Zero-init: at step 0 the prediction is the plain mean of the branches. No new preprocessing. Default `--fusion-combination paper_mean` = previous behavior bit-for-bit (verified on CPU).

- `stanet2`: fusion + fNIRS only (EEG out of the final decision, exactly STA-Net; the EEG branch is still trained by its own loss term).
- `stanet3`: all three branches.

Baseline seed 42: window **0.6074**, trial-level **0.6325**. Also look at the per-window-position accuracy (early windows 0–5 of each block), which is where STA-Net reports its gain.
