# run78 — checkpoint selection (priority-BESTSEL)

**Uses the `-new` scripts** (all three in one folder). Same architecture and data as run60; only HOW the stage-1 checkpoint is chosen changes. No new preprocessing.

**Why (baseline run60, 78 folds):** validation = one trial per class, so val_acc is a few discrete values. corr(val_acc, test_acc) = **−0.04**; val_acc at the chosen epoch averages **0.77** vs test **0.61**; in **25/78 folds the 'best' epoch is ≤ 5**, and those folds score 0.590 vs 0.616 for the rest (weak, possibly confounded evidence).

Baseline seed 42: window **0.6074**, trial-level **0.6325**. Stop rule: ≥ +1 pp AND better in ≥ 16/26 subjects, else reject. Realistic ceiling ≈ +0.8 pp, so judge by the subject count as well as the mean.
