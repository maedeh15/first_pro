```python
import time, datetime

COOLDOWN_MINUTES = 30

print(f"[cooldown] sleeping {COOLDOWN_MINUTES} minutes starting "
      f"{datetime.datetime.now().strftime('%H:%M:%S')} ...")
for remaining in range(COOLDOWN_MINUTES, 0, -1):
    time.sleep(60)
    print(f"  [cooldown] {remaining - 1} minute(s) left "
          f"({datetime.datetime.now().strftime('%H:%M:%S')})")
print(f"[cooldown] done at {datetime.datetime.now().strftime('%H:%M:%S')} -- resuming.")
```
