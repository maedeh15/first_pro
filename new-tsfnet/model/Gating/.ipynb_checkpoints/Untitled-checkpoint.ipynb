```python
import scipy.io as sio
import numpy as np
import os

# مسیر فایل واقعی خودت
mat_path = r"E:\kntu\Thesis\codes\DATA\VP001-EEG\cnt_nback.mat"

# Load
mat = sio.loadmat(mat_path, struct_as_record=False, squeeze_me=True)

# پیدا کردن struct اصلی
keys = [k for k in mat.keys() if not k.startswith("__")]
cnt = mat[keys[0]]

# اطلاعات
data = np.asarray(cnt.x)
ch_names = [str(c).strip() for c in np.asarray(cnt.clab).flatten()]
sfreq = float(cnt.fs)

print("========== EEG FILE CHECK ==========")
print("File:", os.path.basename(mat_path))
print("Data shape:", data.shape)
print("Sampling rate:", sfreq, "Hz")
print("Number of channels:", len(ch_names))

print("\nChannel names:")
for i, ch in enumerate(ch_names):
    print(f"{i:2d}: {ch}")

print("\n---------- Special channels ----------")
print("TP9  exists:", "TP9" in ch_names)
print("TP10 exists:", "TP10" in ch_names)
print("HEOG exists:", "HEOG" in ch_names)
print("VEOG exists:", "VEOG" in ch_names)

print("\n---------- EEG channels ----------")
eeg_channels = [ch for ch in ch_names if ch not in ["HEOG", "VEOG"]]
print("EEG channel count:", len(eeg_channels))
print(eeg_channels)
```
