```python
import os, glob, datetime

base = r"E:\kntu\Thesis\codes\Results\TSFNET_PAPER_EXACT_nback\EGTA"
cutoff = datetime.datetime(2026, 9, 22, 22, 17)

for sub in sorted(os.listdir(base)):
    d = os.path.join(base, sub)
    pts = glob.glob(os.path.join(d, "*.pt"))
    times = [datetime.datetime.fromtimestamp(os.path.getmtime(p)) for p in pts]
    newest, oldest = max(times), min(times)
    after = sum(t > cutoff for t in times)
    logs = [os.path.basename(f) for f in glob.glob(os.path.join(d, "*.txt"))]
    print(f"{sub}")
    print(f"   قدیمی‌ترین: {oldest:%Y-%m-%d %H:%M}   جدیدترین: {newest:%Y-%m-%d %H:%M}")
    print(f"   فایل‌های بعد از ران‌های D1: {after} از {len(pts)}")
    print(f"   لاگ‌های موجود: {logs}")
```
