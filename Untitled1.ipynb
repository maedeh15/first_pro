```python
# ============================================================================
# Cross-subject consistency check: verifies that the findings confirmed on
# VP001 (channel count/names, trial count, sampling rate, amplitude scale,
# spectral content above 50Hz) hold across a SAMPLE of subjects, not just
# VP001. This is a fast, read-only check (no ICA, no filtering) meant to run
# BEFORE the full preprocessing pipeline, to catch any subject whose raw
# data deviates from VP001's structure (different channel set, different
# fs, corrupted markers, etc.) before it silently breaks or skews the run.
#
# Run this on your machine (edit BASE_DATASET_PATH below), then paste the
# printed output back.
# ============================================================================
import os
import numpy as np
import scipy.io as sio

BASE_DATASET_PATH = r"E:\kntu\Thesis\codes\DATA"

# A spread across the cohort, not just consecutive subjects: catches issues
# tied to a specific recording day/batch, not just early vs. late subjects.
SAMPLE_SUBJECTS = [1, 5, 10, 13, 18, 22, 26]

# Reference values, confirmed on VP001 -- every sampled subject is compared
# against these.
EXPECTED_N_CHANNELS = 30
EXPECTED_N_REAL_CHANNELS = 28   # excluding VEOG/HEOG
EXPECTED_FS = 200
EXPECTED_TRIALS = 27            # 9/session x 3 sessions
EXPECTED_CLAB = [
    'Fp1', 'AFF5h', 'AFz', 'F1', 'FC5', 'FC1', 'T7', 'C3', 'Cz', 'CP5',
    'CP1', 'P7', 'P3', 'Pz', 'POz', 'O1', 'Fp2', 'AFF6h', 'F2', 'FC2',
    'FC6', 'C4', 'T8', 'CP2', 'CP6', 'P4', 'P8', 'O2', 'HEOG', 'VEOG',
]


def _load_struct(path):
    mat = sio.loadmat(path, struct_as_record=False, squeeze_me=True)
    keys = [k for k in mat.keys() if not k.startswith("__")]
    return mat[keys[0]]


def high_freq_power_fraction(sig, fs, cutoff=50.0):
    sig = sig - sig.mean()
    freqs = np.fft.rfftfreq(len(sig), d=1.0 / fs)
    power = np.abs(np.fft.rfft(sig)) ** 2
    total = power.sum()
    if total == 0:
        return float("nan")
    return power[freqs >= cutoff].sum() / total


def check_subject(sub_num):
    subject_id = f"VP{sub_num:03d}"
    eeg_folder = os.path.join(BASE_DATASET_PATH, f"{subject_id}-EEG")
    result = {"subject": subject_id, "ok": True, "issues": []}

    try:
        cnt = _load_struct(os.path.join(eeg_folder, "cnt_nback.mat"))
        mrk = _load_struct(os.path.join(eeg_folder, "mrk_nback.mat"))
    except Exception as e:
        result["ok"] = False
        result["issues"].append(f"FAILED TO LOAD: {e}")
        return result

    clab = [str(c).strip() for c in np.asarray(cnt.clab).flatten()]
    fs = float(getattr(cnt, "fs", -1))
    x = np.asarray(cnt.x)
    n_real = len([c for c in clab if c not in ("VEOG", "HEOG")])

    result["n_channels"] = len(clab)
    result["n_real_channels"] = n_real
    result["fs"] = fs
    result["n_samples"] = x.shape[0]
    result["clab_matches_vp001"] = (clab == EXPECTED_CLAB)

    if len(clab) != EXPECTED_N_CHANNELS:
        result["ok"] = False
        result["issues"].append(f"n_channels={len(clab)} (expected {EXPECTED_N_CHANNELS})")
    if not result["clab_matches_vp001"]:
        result["ok"] = False
        result["issues"].append("channel NAMES/ORDER differ from VP001")
    if fs != EXPECTED_FS:
        result["ok"] = False
        result["issues"].append(f"fs={fs} (expected {EXPECTED_FS})")

    # amplitude scale
    absmax = float(np.abs(x[:, 0]).max())
    result["ch0_absmax_uV"] = absmax
    if absmax < 1.0:
        result["ok"] = False
        result["issues"].append(f"amplitude looks like Volts not uV (max={absmax})")

    # NaN/Inf
    n_bad = int(np.isnan(x).sum() + np.isinf(x).sum())
    result["nan_inf_count"] = n_bad
    if n_bad > 0:
        result["ok"] = False
        result["issues"].append(f"{n_bad} NaN/Inf values found")

    # spectral check
    hf_frac = high_freq_power_fraction(x[:, 0], fs)
    result["high_freq_power_fraction"] = hf_frac
    if hf_frac > 0.02:   # VP001 was ~0.0018; flag if clearly different
        result["issues"].append(f">50Hz power fraction={hf_frac:.4f} (VP001 was ~0.0018 -- "
                                 f"this subject may NOT have the same implicit lowpass)")

    # trial markers
    desc = np.asarray(mrk.event.desc, dtype=int).flatten()
    trial_codes = desc[np.isin(desc, [112, 128, 144])]
    result["n_trials"] = len(trial_codes)
    result["trial_class_counts"] = np.bincount(trial_codes)[[112, 128, 144]].tolist() \
        if len(trial_codes) else [0, 0, 0]
    if len(trial_codes) != EXPECTED_TRIALS:
        result["ok"] = False
        result["issues"].append(f"n_trials={len(trial_codes)} (expected {EXPECTED_TRIALS})")

    return result


if __name__ == "__main__":
    print("=" * 90)
    print(f"Checking {len(SAMPLE_SUBJECTS)} subjects against VP001's confirmed reference values")
    print("=" * 90)

    all_results = []
    for sub_num in SAMPLE_SUBJECTS:
        print(f"\nChecking VP{sub_num:03d} ...")
        r = check_subject(sub_num)
        all_results.append(r)
        status = "OK" if r["ok"] else "MISMATCH"
        print(f"  [{status}] {r}")

    print("\n" + "=" * 90)
    print("SUMMARY")
    print("=" * 90)
    n_ok = sum(1 for r in all_results if r["ok"])
    print(f"{n_ok}/{len(all_results)} subjects fully match VP001's reference values.")
    for r in all_results:
        if not r["ok"]:
            print(f"  {r['subject']}: {r['issues']}")

    print("\nCopy/paste this entire output back.")
```
